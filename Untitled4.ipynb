```python
df=pd.read_csv("covid.csv")
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

continent_cases = df.groupby('Continent')['TotalCases'].sum()

plt.figure(figsize=(8,8))
plt.pie(
    continent_cases,
    labels=continent_cases.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Global COVID Cases by Continent')
plt.tight_layout()
plt.show()
```

```python
top_deaths = df.sort_values('Deaths/1M pop', ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.bar(top_deaths['Country/Region'], top_deaths['Deaths/1M pop'])
plt.xticks(rotation=45)
plt.xlabel('Country')
plt.ylabel('Deaths per 1M')
plt.title('Top 10 Countries by COVID Deaths per 1M')
plt.tight_layout()
plt.show()
```

```python
global_recovery = df['TotalRecovered'].sum()
global_deaths = df['TotalDeaths'].sum()

plt.figure(figsize=(6,4))
plt.plot(['Recovered', 'Deaths'], [global_recovery, global_deaths], marker='o')
plt.xlabel('Outcome')
plt.ylabel('Number of People')
plt.title('Global COVID Recovery vs Deaths')
plt.tight_layout()
plt.show()
```
